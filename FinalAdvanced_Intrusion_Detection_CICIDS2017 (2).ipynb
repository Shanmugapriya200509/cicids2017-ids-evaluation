```python
import pandas as pd
import numpy as np

# Maximum samples per class
MAX_SAMPLES_PER_CLASS = 10000

# Dictionary to store selected samples
selected_data = {}

# Track how many samples have been collected
class_counter = {}

print("Creating balanced modeling dataset...")
print("=" * 60)

for chunk in pd.read_csv(
    DATA_FILE,
    encoding="latin1",
    chunksize=100000,
    low_memory=False
):

    # Clean column names
    chunk.columns = chunk.columns.str.strip()

    # Clean labels
    chunk["Label"] = (
        chunk["Label"]
        .astype(str)
        .str.strip()
    )

    # Remove missing labels
    chunk = chunk[
        ~chunk["Label"].isin(
            ["nan", "NaN", "None", ""]
        )
    ]

    # Process each class
    for label in chunk["Label"].unique():

        if label not in selected_data:
            selected_data[label] = []
            class_counter[label] = 0

        remaining = (
            MAX_SAMPLES_PER_CLASS
            - class_counter[label]
        )

        if remaining <= 0:
            continue

        class_rows = chunk[
            chunk["Label"] == label
        ]

        class_rows = class_rows.head(
            remaining
        )

        selected_data[label].append(
            class_rows
        )

        class_counter[label] += len(
            class_rows
        )

    # Stop when every class has reached
    # the maximum available/allowed samples
    if all(
        count >= MAX_SAMPLES_PER_CLASS
        or count == clean_label_counter[label]
        for label, count in class_counter.items()
    ):
        pass

# Combine selected records
balanced_parts = []

for label, parts in selected_data.items():

    if len(parts) > 0:
        balanced_parts.append(
            pd.concat(
                parts,
                ignore_index=True
            )
        )

balanced_data = pd.concat(
    balanced_parts,
    ignore_index=True
)

# Shuffle
balanced_data = balanced_data.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("\nBalanced dataset created!")
print("Shape:", balanced_data.shape)
```

```python
import numpy as np
import pandas as pd

# Make a copy
model_data = balanced_data.copy()

# Clean column names
model_data.columns = model_data.columns.str.strip()

# Convert Destination Port to numeric
model_data["Destination Port"] = pd.to_numeric(
    model_data["Destination Port"],
    errors="coerce"
)

# Clean Label
model_data["Label"] = (
    model_data["Label"]
    .astype(str)
    .str.strip()
)

# Replace infinite values with NaN
model_data = model_data.replace(
    [np.inf, -np.inf],
    np.nan
)

# Remove rows with missing labels
model_data = model_data[
    model_data["Label"].notna()
]

# Remove rows where label became "nan"
model_data = model_data[
    model_data["Label"] != "nan"
]

print("Dataset after initial cleaning:")
print("Rows:", len(model_data))
print("Columns:", len(model_data.columns))
```

```python
X = model_data.drop(
    columns=["Label"]
)

y = model_data["Label"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature data types:")
print(X.dtypes.value_counts())

print("\nNumber of classes:", y.nunique())
print("\nClasses:")
print(sorted(y.unique()))
```

```python
# Convert every feature to numeric
X = X.apply(
    pd.to_numeric,
    errors="coerce"
)

# Replace infinity
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Count missing values
missing_values = X.isna().sum()

print(
    "Total missing feature values:",
    missing_values.sum()
)

print("\nColumns containing missing values:")
print(
    missing_values[
        missing_values > 0
    ]
)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(
    strategy="median"
)

X_clean = imputer.fit_transform(X)

print("Original X shape:", X.shape)
print("Clean X shape:", X_clean.shape)

print(
    "Remaining NaN values:",
    np.isnan(X_clean).sum()
)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Label mapping:")
for i, label in enumerate(label_encoder.classes_):
    print(i, "=", label)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_clean,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
print("Features        :", X_train.shape[1])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training shape:", X_train_scaled.shape)
print("Testing shape :", X_test_scaled.shape)
```

```python
import joblib

joblib.dump(
    scaler,
    "/content/drive/MyDrive/CICIDS2017/scaler.pkl"
)

joblib.dump(
    imputer,
    "/content/drive/MyDrive/CICIDS2017/imputer.pkl"
)

joblib.dump(
    label_encoder,
    "/content/drive/MyDrive/CICIDS2017/label_encoder.pkl"
)

print("Preprocessing objects saved successfully.")
```

```python
from sklearn.ensemble import RandomForestClassifier
import time

print("Training Random Forest...")
print("=" * 60)

start_time = time.time()

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf_model.fit(
    X_train_scaled,
    y_train
)

rf_training_time = time.time() - start_time

print("\nRandom Forest training completed!")
print(
    "Training time:",
    round(rf_training_time, 2),
    "seconds"
)
```

```python
start_time = time.time()

rf_predictions = rf_model.predict(
    X_test_scaled
)

rf_prediction_time = time.time() - start_time

print("Prediction completed!")
print(
    "Prediction time:",
    round(rf_prediction_time, 2),
    "seconds"
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

rf_precision = precision_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)

print("RANDOM FOREST RESULTS")
print("=" * 50)
print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("F1-Score :", round(rf_f1, 4))
```

```python
from sklearn.metrics import classification_report

rf_report = classification_report(
    y_test,
    rf_predictions,
    target_names=label_encoder.classes_,
    zero_division=0
)

print(rf_report)
```

```python
import numpy as np

X_train_lstm = X_train_scaled.reshape(
    X_train_scaled.shape[0],
    X_train_scaled.shape[1],
    1
)

X_test_lstm = X_test_scaled.reshape(
    X_test_scaled.shape[0],
    X_test_scaled.shape[1],
    1
)

print("Bi-LSTM training shape:", X_train_lstm.shape)
print("Bi-LSTM testing shape :", X_test_lstm.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Bidirectional, LSTM, Dense, Dropout

num_classes = len(label_encoder.classes_)

bilstm_model = Sequential()

bilstm_model.add(
    Bidirectional(
        LSTM(64, return_sequences=True),
        input_shape=(78, 1)
    )
)

bilstm_model.add(Dropout(0.3))

bilstm_model.add(
    Bidirectional(
        LSTM(32)
    )
)

bilstm_model.add(Dropout(0.3))

bilstm_model.add(
    Dense(64, activation="relu")
)

bilstm_model.add(Dropout(0.2))

bilstm_model.add(
    Dense(num_classes, activation="softmax")
)

bilstm_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

bilstm_model.summary()
```

```python
import time

print("Training Bi-LSTM...")
print("=" * 60)

start_time = time.time()

history = bilstm_model.fit(
    X_train_lstm,
    y_train,
    validation_split=0.2,
    epochs=10,
    batch_size=256,
    verbose=1
)

bilstm_training_time = time.time() - start_time

print("\nBi-LSTM training completed!")
print(
    "Training time:",
    round(bilstm_training_time, 2),
    "seconds"
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Bi-LSTM Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
start_time = time.time()

bilstm_probabilities = bilstm_model.predict(
    X_test_lstm,
    batch_size=256,
    verbose=1
)

bilstm_prediction_time = time.time() - start_time

bilstm_predictions = np.argmax(
    bilstm_probabilities,
    axis=1
)

print("Bi-LSTM prediction completed!")
print(
    "Prediction time:",
    round(bilstm_prediction_time, 2),
    "seconds"
)
```

```python
from google.colab import files
from PIL import Image, ImageChops

# Upload the 3 ORIGINAL images
uploaded = files.upload()

# Find the images
rf_file = [f for f in uploaded if "RF" in f.upper()][0]
xgb_file = [f for f in uploaded if "XGBOOST" in f.upper()][0]
bilstm_file = [f for f in uploaded if "BILSTM" in f.upper()][0]

# Open original images
rf = Image.open(rf_file).convert("RGB")
xgb = Image.open(xgb_file).convert("RGB")
bilstm = Image.open(bilstm_file).convert("RGB")


# Remove only unnecessary outer white space
def crop_white(img):
    bg = Image.new("RGB", img.size, "white")
    diff = ImageChops.difference(img, bg)

    bbox = diff.getbbox()

    if bbox:
        # Small safety margin
        left = max(0, bbox[0] - 5)
        top = max(0, bbox[1] - 5)
        right = min(img.width, bbox[2] + 5)
        bottom = min(img.height, bbox[3] + 5)

        return img.crop((left, top, right, bottom))

    return img


rf = crop_white(rf)
xgb = crop_white(xgb)
bilstm = crop_white(bilstm)


# ------------------------------------------------
# IMPORTANT:
# Give EACH graph its own fixed-width area.
# This prevents the 3rd graph from being cut.
# ------------------------------------------------

CELL_WIDTH = 1200
CELL_HEIGHT = 1200
GAP = 30

def fit_image(img, max_width, max_height):

    ratio = min(
        max_width / img.width,
        max_height / img.height
    )

    new_width = int(img.width * ratio)
    new_height = int(img.height * ratio)

    resized = img.resize(
        (new_width, new_height),
        Image.Resampling.LANCZOS
    )

    # White background cell
    cell = Image.new(
        "RGB",
        (max_width, max_height),
        "white"
    )

    # Center image inside cell
    x = (max_width - new_width) // 2
    y = (max_height - new_height) // 2

    cell.paste(resized, (x, y))

    return cell


# Fit each graph completely
rf_cell = fit_image(
    rf,
    CELL_WIDTH,
    CELL_HEIGHT
)

xgb_cell = fit_image(
    xgb,
    CELL_WIDTH,
    CELL_HEIGHT
)

bilstm_cell = fit_image(
    bilstm,
    CELL_WIDTH,
    CELL_HEIGHT
)


# ------------------------------------------------
# Create ONE horizontal image
# ------------------------------------------------

final_width = (
    CELL_WIDTH * 3 +
    GAP * 2
)

final_height = CELL_HEIGHT

combined = Image.new(
    "RGB",
    (final_width, final_height),
    "white"
)

# Random Forest
combined.paste(
    rf_cell,
    (0, 0)
)

# XGBoost
combined.paste(
    xgb_cell,
    (CELL_WIDTH + GAP, 0)
)

# Bi-LSTM
combined.paste(
    bilstm_cell,
    (CELL_WIDTH * 2 + GAP * 2, 0)
)


# ------------------------------------------------
# Save high-quality PNG
# ------------------------------------------------

output_file = "FINAL_HORIZONTAL_CONFUSION_MATRICES.png"

combined.save(
    output_file,
    format="PNG",
    dpi=(600, 600),
    optimize=False
)

print("======================================")
print("FINAL IMAGE CREATED")
print("======================================")
print("File:", output_file)
print("Size:", combined.size)

files.download(output_file)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

SEED = 42
CAP = 10000

log = []
def record(stage, d):
    log.append({"stage": stage, "rows": len(d)})
    print(f"{stage}: {len(d):,} rows")

# 1) Load and tidy column names
df = pd.read_csv("combinenew.csv", low_memory=False)
df.columns = df.columns.str.strip()
record("1. Raw merged file", df)

# 2) Remove the blank row and any row with a missing label
df = df.dropna(subset=["Label"])
record("2. After removing blank/unlabelled rows", df)

# 3) Make every feature numeric
feat_cols = [c for c in df.columns if c != "Label"]
df[feat_cols] = df[feat_cols].apply(pd.to_numeric, errors="coerce")

# 4) Remove exact duplicate rows
df = df.drop_duplicates()
record("3. After removing exact duplicates", df)

# 5) Remove rows whose features are identical but labels differ (label noise)
conflict = df.duplicated(subset=feat_cols, keep=False)
print("   conflicting rows removed:", int(conflict.sum()))
df = df[~conflict]
record("4. After removing conflicting-label rows", df)

# 6) Remove rows with inf / NaN feature values
df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=feat_cols)
record("5. After removing inf/NaN rows", df)
print("\nClass distribution after cleaning:\n", df["Label"].value_counts())

# 7) Cap each class at CAP rows (classes smaller than CAP keep all rows)
parts = []
for lab, g in df.groupby("Label"):
    parts.append(g.sample(n=min(len(g), CAP), random_state=SEED))
capped = pd.concat(parts).reset_index(drop=True)
record("6. After capping each class at 10,000", capped)
print("\nClass distribution after cap:\n", capped["Label"].value_counts())

# 8) Stratified 80/20 split
train_df, test_df = train_test_split(
    capped, test_size=0.20, random_state=SEED, stratify=capped["Label"]
)
record("7a. Training set", train_df)
record("7b. Test set", test_df)

# 9) Check: no identical rows shared between train and test
overlap = train_df.merge(test_df, on=feat_cols, how="inner")
print("\nRows shared between train and test:", len(overlap))

# 10) Save everything for the next steps
train_df.to_csv("clean_train.csv", index=False)
test_df.to_csv("clean_test.csv", index=False)
pd.DataFrame(log).to_csv("pipeline_log.csv", index=False)
print("\nSaved: clean_train.csv, clean_test.csv, pipeline_log.csv")
```

```python
import numpy as np, pandas as pd, time
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             matthews_corrcoef, confusion_matrix,
                             roc_auc_score, average_precision_score)

# Rejoin the clean capped data; each seed makes a fresh stratified 80/20 split
full = pd.concat([pd.read_csv("clean_train.csv"),
                  pd.read_csv("clean_test.csv")], ignore_index=True)
feat_cols = [c for c in full.columns if c != "Label"]
le = LabelEncoder()
X = full[feat_cols].values
y = le.fit_transform(full["Label"])
n_cls = len(le.classes_)
print("Rows:", len(full), "| Features:", len(feat_cols), "| Classes:", n_cls)

def macro_fpr_fnr(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred, labels=list(range(n_cls)))
    tp = np.diag(cm); fn = cm.sum(1) - tp; fp = cm.sum(0) - tp
    tn = cm.sum() - tp - fn - fp
    return (fp / (fp + tn)).mean(), (fn / (fn + tp)).mean()

def evaluate(y_te, proba):
    proba = proba / proba.sum(axis=1, keepdims=True)
    pred = proba.argmax(axis=1)
    fpr, fnr = macro_fpr_fnr(y_te, pred)
    y_bin = label_binarize(y_te, classes=list(range(n_cls)))
    return {
        "accuracy": accuracy_score(y_te, pred),
        "balanced_acc": balanced_accuracy_score(y_te, pred),
        "macro_f1": f1_score(y_te, pred, average="macro"),
        "weighted_f1": f1_score(y_te, pred, average="weighted"),
        "mcc": matthews_corrcoef(y_te, pred),
        "macro_fpr": fpr,
        "macro_fnr": fnr,
        "roc_auc": roc_auc_score(y_te, proba, multi_class="ovr", average="macro"),
        "pr_auc": average_precision_score(y_bin, proba, average="macro"),
    }

SEEDS = list(range(42, 52))
rows = []
for seed in SEEDS:
    t0 = time.time()
    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=0.20, random_state=seed, stratify=y)

    rf = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=seed)
    rf.fit(X_tr, y_tr)
    rows.append({"seed": seed, "model": "RandomForest",
                 **evaluate(y_te, rf.predict_proba(X_te))})

    xgb = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                        objective="multi:softprob", tree_method="hist",
                        n_jobs=-1, random_state=seed)
    xgb.fit(X_tr, y_tr)
    rows.append({"seed": seed, "model": "XGBoost",
                 **evaluate(y_te, xgb.predict_proba(X_te))})
    print(f"seed {seed} done in {time.time()-t0:.0f}s")

res = pd.DataFrame(rows)
res.to_csv("results_rf_xgb.csv", index=False)

metrics = ["accuracy", "balanced_acc", "macro_f1", "weighted_f1",
           "mcc", "macro_fpr", "macro_fnr", "roc_auc", "pr_auc"]

print("\n=== Mean ± SD over", len(SEEDS), "seeds ===")
for m_name, g in res.groupby("model"):
    print("\n" + m_name)
    for m in metrics:
        print(f"  {m:13s} {g[m].mean():.4f} ± {g[m].std():.4f}")

print("\n=== Paired t-test, RandomForest vs XGBoost ===")
rf_r = res[res.model == "RandomForest"].sort_values("seed")
xg_r = res[res.model == "XGBoost"].sort_values("seed")
for m in ["accuracy", "macro_f1", "mcc"]:
    t, p = stats.ttest_rel(rf_r[m].values, xg_r[m].values)
    print(f"  {m:10s} t = {t:.3f}, p = {p:.4f}")

print("\nSaved: results_rf_xgb.csv")
```

```python
import numpy as np, pandas as pd, time
import tensorflow as tf
from scipy import stats
from tensorflow.keras import layers, models, callbacks
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler, label_binarize
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             matthews_corrcoef, confusion_matrix,
                             roc_auc_score, average_precision_score)

full = pd.concat([pd.read_csv("clean_train.csv"),
                  pd.read_csv("clean_test.csv")], ignore_index=True)
feat_cols = [c for c in full.columns if c != "Label"]
le = LabelEncoder()
X = full[feat_cols].values.astype("float32")
y = le.fit_transform(full["Label"])
n_cls = len(le.classes_)

def macro_fpr_fnr(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred, labels=list(range(n_cls)))
    tp = np.diag(cm); fn = cm.sum(1) - tp; fp = cm.sum(0) - tp
    tn = cm.sum() - tp - fn - fp
    return (fp / (fp + tn)).mean(), (fn / (fn + tp)).mean()

def evaluate(y_te, proba):
    proba = proba / proba.sum(axis=1, keepdims=True)
    pred = proba.argmax(axis=1)
    fpr, fnr = macro_fpr_fnr(y_te, pred)
    y_bin = label_binarize(y_te, classes=list(range(n_cls)))
    return {
        "accuracy": accuracy_score(y_te, pred),
        "balanced_acc": balanced_accuracy_score(y_te, pred),
        "macro_f1": f1_score(y_te, pred, average="macro"),
        "weighted_f1": f1_score(y_te, pred, average="weighted"),
        "mcc": matthews_corrcoef(y_te, pred),
        "macro_fpr": fpr, "macro_fnr": fnr,
        "roc_auc": roc_auc_score(y_te, proba, multi_class="ovr", average="macro"),
        "pr_auc": average_precision_score(y_bin, proba, average="macro"),
    }

def build_model():
    m = models.Sequential([
        layers.Input(shape=(X.shape[1], 1)),
        layers.Bidirectional(layers.LSTM(64)),
        layers.Dropout(0.3),
        layers.Dense(64, activation="relu"),
        layers.Dense(n_cls, activation="softmax"),
    ])
    m.compile(optimizer="adam", loss="sparse_categorical_crossentropy")
    return m

SEEDS = list(range(42, 52))
rows = []
for seed in SEEDS:
    t0 = time.time()
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=0.20, random_state=seed, stratify=y)

    scaler = StandardScaler().fit(X_tr)          # fitted on training data only
    X_tr = np.clip(scaler.transform(X_tr), -10, 10)
    X_te = np.clip(scaler.transform(X_te), -10, 10)
    X_tr = X_tr[..., None]; X_te = X_te[..., None]

    model = build_model()
    es = callbacks.EarlyStopping(monitor="val_loss", patience=3,
                                 restore_best_weights=True)
    model.fit(X_tr, y_tr, validation_split=0.10, epochs=30,
              batch_size=256, callbacks=[es], verbose=0)

    proba = model.predict(X_te, batch_size=1024, verbose=0)
    rows.append({"seed": seed, "model": "BiLSTM", **evaluate(y_te, proba)})
    print(f"seed {seed} done in {time.time()-t0:.0f}s")

res = pd.DataFrame(rows)
res.to_csv("results_bilstm.csv", index=False)

metrics = ["accuracy", "balanced_acc", "macro_f1", "weighted_f1",
           "mcc", "macro_fpr", "macro_fnr", "roc_auc", "pr_auc"]
print("\n=== BiLSTM: Mean ± SD over", len(SEEDS), "seeds ===")
for m in metrics:
    print(f"  {m:13s} {res[m].mean():.4f} ± {res[m].std():.4f}")

# Paired t-tests against the other two models
old = pd.read_csv("results_rf_xgb.csv")
print("\n=== Paired t-test, BiLSTM vs each model ===")
for other in ["RandomForest", "XGBoost"]:
    o = old[old.model == other].sort_values("seed")
    b = res.sort_values("seed")
    for m in ["accuracy", "macro_f1", "mcc"]:
        t, p = stats.ttest_rel(b[m].values, o[m].values)
        print(f"  BiLSTM vs {other:12s} {m:9s} t = {t:.3f}, p = {p:.4f}")

print("\nSaved: results_bilstm.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.preprocessing import label_binarize
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             matthews_corrcoef, confusion_matrix,
                             average_precision_score)

def evaluate(y_te, proba):
    pred = proba.argmax(axis=1)
    cm = confusion_matrix(y_te, pred, labels=list(range(n_cls)))
    tp = np.diag(cm); fn = cm.sum(1) - tp
    y_bin = label_binarize(y_te, classes=list(range(n_cls)))
    return {
        "accuracy": accuracy_score(y_te, pred),
        "balanced_acc": balanced_accuracy_score(y_te, pred),
        "macro_f1": f1_score(y_te, pred, average="macro"),
        "mcc": matthews_corrcoef(y_te, pred),
        "macro_fnr": (fn / (fn + tp)).mean(),
        "pr_auc": average_precision_score(y_bin, proba, average="macro"),
        "benign_false_alarm": (pred[y_te == benign] != benign).mean(),
        "attacks_missed_as_benign": (pred[y_te != benign] == benign).mean(),
    }

SEEDS = [42, 43, 44]
CONDS = [("cap5k", 5000, False),
         ("cap10k", 10000, False),
         ("cap10k+weights", 10000, True)]
PART = "ablation_partial.csv"
rows = pd.read_csv(PART).to_dict("records") if os.path.exists(PART) else []
done = {(int(r["seed"]), r["condition"], r["model"]) for r in rows}

for seed in SEEDS:
    tr_idx, te_idx = train_test_split(np.arange(len(y)), test_size=0.20,
                                      random_state=seed, stratify=y)
    X_te, y_te = X[te_idx], y[te_idx]
    for cond, cap, use_w in CONDS:
        if (seed, cond, "RandomForest") in done and (seed, cond, "XGBoost") in done:
            continue
        t0 = time.time()
        rng = np.random.RandomState(seed)
        keep = []
        for c in range(n_cls):
            idx = tr_idx[y[tr_idx] == c]
            if len(idx) > cap:
                idx = rng.choice(idx, cap, replace=False)
            keep.append(idx)
        keep = np.concatenate(keep)
        X_tr, y_tr = X[keep], y[keep]
        sw = compute_sample_weight("balanced", y_tr) if use_w else None

        rf = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=seed)
        rf.fit(X_tr, y_tr, sample_weight=sw)
        rows.append({"seed": seed, "condition": cond, "model": "RandomForest",
                     "train_rows": len(y_tr), **evaluate(y_te, rf.predict_proba(X_te))})

        xgb = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                            objective="multi:softprob", tree_method="hist",
                            n_jobs=-1, random_state=seed)
        xgb.fit(X_tr, y_tr, sample_weight=sw)
        rows.append({"seed": seed, "condition": cond, "model": "XGBoost",
                     "train_rows": len(y_tr), **evaluate(y_te, xgb.predict_proba(X_te))})

        pd.DataFrame(rows).to_csv(PART, index=False)   # save after every condition
        print(f"seed {seed} | {cond} done in {time.time()-t0:.0f}s")

res = pd.DataFrame(rows)
res.to_csv("ablation_results.csv", index=False)

metrics = ["accuracy", "balanced_acc", "macro_f1", "mcc", "macro_fnr",
           "pr_auc", "benign_false_alarm", "attacks_missed_as_benign"]
table = {}
for (cond, model), g in res.groupby(["condition", "model"]):
    table[(cond, model)] = {m: f"{g[m].mean():.4f} ± {g[m].std():.4f}" for m in metrics}
print("\n=== Mean ± SD over", len(SEEDS), "seeds (same test set for all conditions) ===")
print(pd.DataFrame(table).T.to_string())

from google.colab import files
files.download("ablation_results.csv")
```

```python
import numpy as np, pandas as pd, time
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import balanced_accuracy_score, f1_score, matthews_corrcoef

CAP = 10000

df = pd.read_csv("combinenew.csv", low_memory=False)
df.columns = df.columns.str.strip()
df = df.dropna(subset=["Label"])
feat_cols = [c for c in df.columns if c != "Label"]
df[feat_cols] = df[feat_cols].apply(pd.to_numeric, errors="coerce")
print("Labelled rows:", len(df))

def cap_sample(d):
    parts = [g.sample(n=min(len(g), CAP), random_state=42)
             for _, g in d.groupby("Label")]
    return pd.concat(parts).reset_index(drop=True)

# BEFORE: duplicates kept, median imputation as in the original paper
medians = {}
for c in feat_cols:
    s = df[c].to_numpy()
    medians[c] = np.median(s[np.isfinite(s)])
before = cap_sample(df)
before[feat_cols] = before[feat_cols].replace([np.inf, -np.inf], np.nan).fillna(pd.Series(medians))
print("BEFORE (duplicates kept):", len(before), "rows (expected 79,323)")

# AFTER: the clean pipeline from Step 3
d = df.drop_duplicates()
d = d[~d.duplicated(subset=feat_cols, keep=False)]
d = d.replace([np.inf, -np.inf], np.nan).dropna(subset=feat_cols)
after = cap_sample(d)
print("AFTER (deduplicated):", len(after), "rows (expected 73,899)")
del df, d

le = LabelEncoder().fit(before["Label"])
def prep(d):
    return (d[feat_cols].values, le.transform(d["Label"]),
            pd.util.hash_pandas_object(d[feat_cols], index=False).values)
data = {"with duplicates": prep(before), "deduplicated": prep(after)}
del before, after

def models(seed):
    return [("RandomForest", RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=seed)),
            ("XGBoost", XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                                      objective="multi:softprob", tree_method="hist",
                                      n_jobs=-1, random_state=seed))]

rows = []
for seed in range(42, 52):
    t0 = time.time()
    for cond, (X, y, h) in data.items():
        tr, te = train_test_split(np.arange(len(y)), test_size=0.20,
                                  random_state=seed, stratify=y)
        leaked = np.isin(h[te], h[tr])
        for name, m in models(seed):
            m.fit(X[tr], y[tr])
            p = m.predict(X[te])
            ok = (p == y[te])
            rows.append({"seed": seed, "condition": cond, "model": name,
                         "accuracy": ok.mean(),
                         "balanced_acc": balanced_accuracy_score(y[te], p),
                         "macro_f1": f1_score(y[te], p, average="macro"),
                         "mcc": matthews_corrcoef(y[te], p),
                         "test_rows_in_train": leaked.mean(),
                         "acc_on_seen_rows": ok[leaked].mean() if leaked.any() else np.nan,
                         "acc_on_unseen_rows": ok[~leaked].mean()})
    pd.DataFrame(rows).to_csv("dedup_before_after_partial.csv", index=False)
    print(f"seed {seed} done in {time.time()-t0:.0f}s")

res = pd.DataFrame(rows)
res.to_csv("dedup_before_after.csv", index=False)

cols = ["accuracy", "balanced_acc", "macro_f1", "mcc",
        "test_rows_in_train", "acc_on_seen_rows", "acc_on_unseen_rows"]
for (cond, model), g in res.groupby(["condition", "model"]):
    print(f"\n{cond} | {model}")
    for c in cols:
        print(f"  {c:20s} {g[c].mean():.4f} ± {g[c].std():.4f}")

print("\n=== Paired t-test across seeds: with duplicates vs deduplicated ===")
for model in ["RandomForest", "XGBoost"]:
    a = res[(res.condition == "with duplicates") & (res.model == model)].sort_values("seed")
    b = res[(res.condition == "deduplicated") & (res.model == model)].sort_values("seed")
    for mt in ["accuracy", "macro_f1", "mcc"]:
        t, p = stats.ttest_rel(a[mt].values, b[mt].values)
        print(f"  {model:13s} {mt:10s} t = {t:.3f}, p = {p:.4f}")

from google.colab import files
files.download("dedup_before_after.csv")
```

```python
import pandas as pd, numpy as np, os, time
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score
from sklearn.metrics import precision_recall_fscore_support, matthews_corrcoef
from sklearn.metrics import confusion_matrix, roc_auc_score
from sklearn.metrics import average_precision_score, classification_report
from xgboost import XGBClassifier

cap = pd.read_csv("clean_capped.csv")
feats = [c for c in cap.columns if c != "Label"]
le = LabelEncoder()
y = le.fit_transform(cap["Label"])
X = cap[feats].astype("float64")
classes = list(le.classes_)
K = len(classes)
benign = classes.index("BENIGN")
print("classes:", K)


def metrics(yt, yp, proba):
    cm = confusion_matrix(yt, yp, labels=range(K))
    tp = np.diag(cm)
    fp = cm.sum(0) - tp
    fn = cm.sum(1) - tp
    tn = cm.sum() - tp - fp - fn
    fpr = np.mean(fp / (fp + tn))
    fnr = np.mean(fn / (fn + tp))
    pm, rm, fm, _ = precision_recall_fscore_support(
        yt, yp, average="macro", zero_division=0)
    pw, rw, fw, _ = precision_recall_fscore_support(
        yt, yp, average="weighted", zero_division=0)
    Y = label_binarize(yt, classes=range(K))
    atk = yt != benign
    pr = [average_precision_score(Y[:, k], proba[:, k]) for k in range(K)]
    return dict(
        accuracy=accuracy_score(yt, yp),
        balanced_acc=balanced_accuracy_score(yt, yp),
        macro_p=pm, macro_r=rm, macro_f1=fm,
        weighted_p=pw, weighted_r=rw, weighted_f1=fw,
        mcc=matthews_corrcoef(yt, yp),
        macro_fpr=fpr, macro_fnr=fnr,
        roc_auc=roc_auc_score(Y, proba, average="macro", multi_class="ovr"),
        pr_auc=np.mean(pr),
        benign_false_alarm=np.mean(yp[yt == benign] != benign),
        attacks_missed_as_benign=np.mean(yp[atk] == benign))

print("Cell A done")
```
